# Basic Translation Example

```python
from transformers import AutoTokenizer, AutoModelForCausalLM
import torch

# Load the pre-trained model and tokenizer
# Using the Facebook Incoder model which is good at code translation
model_name = "facebook/incoder-1B"
tokenizer = AutoTokenizer.from_pretrained(model_name)
model = AutoModelForCausalLM.from_pretrained(model_name)

# This is our simple add function in Python that we'll translate
source_code = '''
# Translate this Python function to Java
def add(a, b):
    return a + b
# Java version:
'''

# Convert the text to tokens the model can understand
inputs = tokenizer(source_code, return_tensors="pt")

# Remove any token_type_ids if present (not needed for generation)
inputs = {k: v for k, v in inputs.items() if k != "token_type_ids"}

# Generate the translated code (256 is plenty of tokens for this simple function)
outputs = model.generate(**inputs, max_length=256)

# Convert back from tokens to text
result = tokenizer.decode(outputs[0], skip_special_tokens=True)

# Show the result
print("Translated Code:\n")
print(result)
```

# Interactive Translation
user choose which language to translate to:

```python
from transformers import AutoTokenizer, AutoModelForCausalLM
import torch

# Load our model and tokenizer (same as before)
model_name = "facebook/incoder-1B"
tokenizer = AutoTokenizer.from_pretrained(model_name)
model = AutoModelForCausalLM.from_pretrained(model_name)

# Ask the user what language they want
target_language = input("Enter target language (e.g., Java, C++, JavaScript): ")

# Python function we're going to translate - a simple factorial function
python_function = '''
def factorial(n):
    if n == 0:
        return 1
    else:
        return n * factorial(n - 1)
'''

# Create a prompt that tells the model what to do
source_code = f'''
# Translate this Python function to {target_language}
{python_function}
# {target_language} version:
'''

# Process the input text
inputs = tokenizer(source_code, return_tensors="pt")
inputs = {k: v for k, v in inputs.items() if k != "token_type_ids"}

# Generate translation with the same settings as before
outputs = model.generate(**inputs, max_length=256, num_return_sequences=1)

# Show the result with a nice checkmark
result = tokenizer.decode(outputs[0], skip_special_tokens=True)
print(f"\n✅ Translated Code to {target_language}:\n")
print(result)
```

# Multi-Language Translation
Translate the same function to multiple languages at once

```python
from transformers import AutoTokenizer, AutoModelForCausalLM
import torch

# Load model and tokenizer as usual
model_name = "facebook/incoder-1B"
tokenizer = AutoTokenizer.from_pretrained(model_name)
model = AutoModelForCausalLM.from_pretrained(model_name)

# List of languages we want to translate to
target_languages = ["Java", "C++", "JavaScript", "C#", "Go", "Rust"]

# Our Python factorial function (same as before)
python_function = '''
def factorial(n):
    if n == 0:
        return 1
    else:
        return n * factorial(n - 1)
'''

# Loop through each language and translate the function
for target_language in target_languages:
    # Show a rotating icon to indicate work in progress
    print(f"\n🔁 Translating to {target_language}...\n")

    # Create the prompt for this language
    source_code = f'''
# Translate this Python function to {target_language}
{python_function}
# {target_language} version:
'''

    # Process input
    inputs = tokenizer(source_code, return_tensors="pt")
    inputs = {k: v for k, v in inputs.items() if k != "token_type_ids"}

    # Generate the translated code
    outputs = model.generate(**inputs, max_length=256, num_return_sequences=1)

    # Show just the relevant part of the translation
    result = tokenizer.decode(outputs[0], skip_special_tokens=True)
    print(f"✅ Translated Code to {target_language}:\n")
    
    # Extract just the translated portion and clean up the output
    translated_part = result.split(f"# {target_language} version:")[-1].strip()
    print(translated_part)
    
    # Add a divider between languages
    print("-" * 50)
```
